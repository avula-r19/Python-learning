```python
#q1


contacts = {}

while True:
    print("\n1. Add contact")
    print("2. Search contact")
    print("3. Show all contacts")
    print("4. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        name = input("Enter name: ")
        phone = input("Enter phone number: ")

        if name in contacts:
            print("Contact already exists")
        else:
            contacts[name] = phone
            print("Contact added")

    elif choice == "2":
        name = input("Enter name to search: ")

        if name in contacts:
            print("Phone number:", contacts[name])
        else:
            print("Contact not found")

    elif choice == "3":
        for name, phone in contacts.items():
            print(name, ":", phone)

    elif choice == "4":
        print("Exiting contact book")
        break

    else:
        print("Invalid choice")
```

```python
#q3

marks = {
    "Aman": 82,
    "Priya": 91,
    "Rahul": 45,
    "Sneha": 67,
    "Kiran": 30
}

passed_students = {}

for name, mark in marks.items():
    if mark >= 40:
        print(name, "Pass")
        passed_students[name] = mark
    else:
        print(name, "Fail")

topper = max(marks, key=marks.get)

print("Topper:", topper)
print("Topper marks:", marks[topper])
print("Passed students:", passed_students)
```

```python
#q4 

riya_friends = {"Aman", "Neha", "Karan", "Sita", "Vikram"}
arjun_friends = {"Neha", "Vikram", "Pooja", "Aman", "Rohan"}

mutual = riya_friends.intersection(arjun_friends)

only_riya = riya_friends - arjun_friends

all_friends = riya_friends.union(arjun_friends)

suggestions = arjun_friends - riya_friends

print("Mutual friends:", mutual)
print("Only Riya's friends:", only_riya)
print("All friends:", all_friends)
print("Friend suggestions for Riya:", suggestions)
```

```python
#q5

text = input("Enter text: ")

char_count = {}

for char in text:
    if char != " ":
        if char in char_count:
            char_count[char] += 1
        else:
            char_count[char] = 1

print(char_count)

if len(char_count) > 0:
    most_frequent = max(char_count, key=char_count.get)

    print("Most frequent character:", most_frequent)
    print("Count:", char_count[most_frequent])
else:
    print("No characters to count")
```

```python
#q6

menu = {
    "dosa": 60,
    "idli": 40,
    "vada": 30,
    "coffee": 25,
    "tea": 15
}

total = 0

while True:
    item = input("Enter item (or bill): ").lower()

    if item == "bill":
        break

    if item in menu:
        total += menu[item]
        print("Added:", item)
    else:
        print("Item not available")

print("Total before discount:", total)

if total > 200:
    discount = total * 0.10
    total = total - discount
    print("Discount applied:", discount)

print("Final bill:", total)
```

```python
#q7

cities = [
    ("Hyderabad", 17.38, 78.48),
    ("Mumbai", 19.07, 72.87),
    ("Chennai", 13.08, 80.27),
    ("Delhi", 28.61, 77.20)
]

highest_latitude = 0
northern_city = ""

for city, latitude, longitude in cities:
    print(city, "Latitude:", latitude, "Longitude:", longitude)

    if latitude > highest_latitude:
        highest_latitude = latitude
        northern_city = city

print("Farthest north city:", northern_city)
print("Highest latitude:", highest_latitude)

#cities[1][1] = 20.00
# TypeError: 'tuple' object does not support item assignment
# Reason: Tuples are immutable, so their values cannot be changed.
```

```python
#q8

books = {
    "Python Basics": 3,
    "Data Science 101": 0,
    "Machine Learning": 2,
    "SQL Guide": 1
}

book_name = input("Enter book name: ")

if book_name not in books:
    print("Book not found")

elif books[book_name] == 0:
    print("Currently unavailable")

else:
    books[book_name] -= 1
    print("Book issued successfully")

print("Updated books:", books)
```

```python
#q9

employees = [
    ("Ravi", "IT"),
    ("Anjali", "HR"),
    ("Suresh", "IT"),
    ("Meena", "Finance"),
    ("Arun", "HR"),
    ("Divya", "IT")
]

departments = {}

for name, department in employees:
    if department not in departments:
        departments[department] = []

    departments[department].append(name)

print(departments)

for department, names in departments.items():
    print(department, ":", len(names), "employees")
```

```python
#q10

answer_key = ("B", "C", "A", "D", "B")

score = 0
wrong_questions = set()

for i in range(5):
    answer = input(f"Enter answer for Q{i + 1}: ").upper()

    if answer == answer_key[i]:
        print("Question", i + 1, "Correct")
        score += 1
    else:
        print("Question", i + 1, "Wrong")
        wrong_questions.add(i + 1)

print("Wrong questions:", sorted(wrong_questions))
print("Final score:", score, "/ 5")

if score == 5:
    print("Excellent")
elif score >= 3:
    print("Good")
else:
    print("Needs improvement")
```

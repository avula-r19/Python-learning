```python
#q1 

marks = [78, 45, 92, 33, 67, 88, 51, 29]

highest = max(marks)
lowest = min(marks)
average = sum(marks) / len(marks)

passed = 0

for mark in marks:
    if mark >= 35:
        passed += 1

print("Highest marks:", highest)
print("Lowest marks:", lowest)
print("Class average:", average)
print("Students passed:", passed)
```

```python
#q2

shopping_list = []

while True:
    item = input("Enter item: ").strip().lower()

    if item == "done":
        break
    elif item in shopping_list:
        print("Already added")
    else:
        shopping_list.append(item)

for i in range(len(shopping_list)):
    print(f"{i + 1}. {shopping_list[i]}")

print("Total items:", len(shopping_list))
```

```python
#q3

username = input("Enter username: ")

if username == "":
    print("Invalid: Username cannot be empty")
elif username[0].isdigit():
    print("Invalid: Username cannot start with a digit")
elif len(username) < 5 or len(username) > 12:
    print("Invalid: Username must have 5 to 12 characters")
elif " " in username:
    print("Invalid: Username cannot contain spaces")
else:
    print("Valid username")
```

```python
#q4

sentence = input("Enter a sentence: ").lower()

vowels = 0
consonants = 0
spaces = 0

for char in sentence:
    if char in "aeiou":
        vowels += 1
    elif char >= "a" and char <= "z":
        consonants += 1
    elif char == " ":
        spaces += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Spaces:", spaces)
```

```python
#q6


temps = [32, 35, 38, 41, 36, 29, 40]
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

fahrenheit = []

for i in range(len(temps)):
    if temps[i] > 37:
        print("Heat alert:", days[i])

    f = temps[i] * 9 / 5 + 32
    fahrenheit.append(f)

hottest = max(temps)
hottest_day = days[temps.index(hottest)]

print("Hottest day:", hottest_day)
print("Hottest temperature:", hottest)
print("Fahrenheit temperatures:", fahrenheit)
```

```python
#q7


sentence = input("Enter a sentence: ")

words = sentence.split()

if len(words) == 0:
    print("No words entered")
else:
    longest = words[0]
    shortest = words[0]
    capital_count = 0

    for word in words:
        if len(word) > len(longest):
            longest = word

        if len(word) < len(shortest):
            shortest = word

        if word[0].isupper():
            capital_count += 1

    print("Longest word:", longest)
    print("Shortest word:", shortest)
    print("Words starting with capital letter:", capital_count)
```

```python
#q8


balances = [1500, -200, 3000, -50, 0, 750, -1000]

positive_accounts = []
negative_accounts = []

for balance in balances:
    if balance > 0:
        positive_accounts.append(balance)
    elif balance < 0:
        negative_accounts.append(balance)

print("Positive accounts:", positive_accounts)
print("Negative accounts:", negative_accounts)
print("Total amount owed:", -sum(negative_accounts))
```

```python
#q9

password = input("Enter password: ")

upper = False
lower = False
digit = False

for char in password:
    if char.isupper():
        upper = True
    elif char.islower():
        lower = True
    elif char.isdigit():
        digit = True

conditions = 0

if upper:
    conditions += 1

if lower:
    conditions += 1

if digit:
    conditions += 1

if len(password) >= 8:
    conditions += 1

if conditions == 4:
    print("Strong")
elif conditions >= 2:
    print("Medium")
else:
    print("Weak")
```

```python
#q10

full_name = input("Enter full name: ").strip()

names = full_name.split()

if len(names) == 0:
    print("Please enter a name")
else:
    initials = ""
    short_name = ""

    for name in names:
        initials += name[0].upper() + "."

    for name in names[:-1]:
        short_name += name[0].upper() + ". "

    short_name += names[-1].capitalize()

    print("Initials:", initials)
    print("Title case:", full_name.title())
    print("Short name:", short_name)
```

```python
# Q1 - Movie Ticket Counter

age = int(input("Enter your age: "))

if age < 0:
    print("Invalid age")
elif age < 5:
    print("Free ticket")
elif age < 18:
    print("Ticket price is Rs. 100")
elif age < 60:
    print("Ticket price is Rs. 200")
else:
    print("Ticket price is Rs. 120")
```

```python
# Q2 - Electricity Bill Calculator

units = int(input("Enter electricity units: "))

if units < 0:
    print("Invalid units")
elif units <= 100:
    bill = units * 3
    print("Electricity bill:", bill)
elif units <= 200:
    bill = 100 * 3 + (units - 100) * 5
    print("Electricity bill:", bill)
else:
    bill = 100 * 3 + 100 * 5 + (units - 200) * 8
    print("Electricity bill:", bill)
```

```python
# Q3 - Multiplication Table

number = int(input("Enter a number: "))

for i in range(1, 11):
    print(number, "x", i, "=", number * i)
```

```python
# Q4 - Savings Piggy Bank

total = 0
days = 0

while total < 1000:
    savings = int(input("Enter today's savings: "))

    if savings <= 0:
        print("Enter a positive amount")
        continue

    total = total + savings
    days = days + 1

print("Savings goal reached!")
print("Total savings:", total)
print("Number of days:", days)
```

```python
# Q5 - ATM PIN Check

correct_pin = 4321
attempts = 0

while attempts < 3:
    pin = int(input("Enter your PIN: "))

    if pin == correct_pin:
        print("Access granted")
        break
    else:
        attempts = attempts + 1
        print("Incorrect PIN")

if attempts == 3:
    print("Card blocked")
```

```python
# Q6 - Even and Odd Counter

n = int(input("Enter N: "))

even = 0
odd = 0

for i in range(1, n + 1):
    if i % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("Even count:", even)
print("Odd count:", odd)
```

```python
# Q7 - Number Guessing Game

secret_number = 27
count = 0

while True:
    guess = int(input("Guess the number: "))
    count = count + 1

    if guess > secret_number:
        print("Too high")
    elif guess < secret_number:
        print("Too low")
    else:
        print("Correct guess!")
        print("Number of guesses:", count)
        break
```

```python
# Q8 - Exam Result with Grade

marks = int(input("Enter your marks: "))

if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
elif marks >= 35:
    print("Grade D")
else:
    print("Fail")
```

```python
# Q9 - Sum of Digits

number = int(input("Enter bill number: "))

if number < 0:
    print("Invalid bill number")
else:
    total = 0

    while number > 0:
        digit = number % 10
        total = total + digit
        number = number // 10

    print("Sum of digits:", total)

    if total > 20:
        print("You won a free gift!")
    else:
        print("No free gift")
```

```python
# Q10 - Staircase Pattern

steps = int(input("Enter number of steps: "))

for i in range(1, steps + 1):
    for j in range(i):
        print("*", end=" ")
    print()
```
